```r
FCs_day04=read.csv("FCs_day04.csv", row.names = 1)
FDRs_day04=read.csv("FDRs_day04.csv", row.names = 1)
FCs_day10=read.csv("FCs_day10.csv", row.names = 1)
FDRs_day10=read.csv("FDRs_day10.csv", row.names = 1)
```

```r
a=data.frame(read.csv("GeneClusters.csv", row.names = 1))
```

```r
a$gene_id  = sapply(a$gene_id , function(x){ gsub("-", ".", x, fixed = TRUE)})
rownames(a) = a$gene_id
```

```r
FCs_day10=FCs_day10[,a$gene_id]
FCs_day04=FCs_day04[,a$gene_id]
```

```r
options(repr.plot.width = 10, repr.plot.height = 20)

tmp_fc = copy(FCs_day04)
tmp_fdr = FDRs_day04

tmp_fc[tmp_fdr > 0.1] = 1
tmp_fc = tmp_fc-1

tmp_fc[tmp_fc > 1] = 1
tmp_fc[tmp_fc < -1] = -1

annotCol = a[,c("pertIndex", "leiden")]
annotCol$leiden = as.factor(annotCol$leiden )
annotCol$pertIndex = NULL
annotCol$gene_id = NULL
    



k <- pheatmap(tmp_fc,  
         fontsize_row = 10,
         fontsize_col = 1,
         cluster_rows=TRUE, 
         cluster_cols=FALSE, 
         annotation_col = annotCol,
         color=colorRampPalette(c("navy", "white", "red"))(50))
#save_pheatmap_pdf(k, filename=paste0(cellLine,"_betaMatrix.pdf"), width=9, height=7)

# options(repr.plot.width=12, repr.plot.height=10)
# covGenes = data.frame(cov(tmp_fc))
# rownames(covGenes) = colnames(tmp_fc)
# colnames(covGenes) = colnames(tmp_fc)

# covGenes[covGenes > 0.04] = 0.04
# covGenes[covGenes < -0.04] = -0.04

# k <- pheatmap(covGenes, treeheight_col=0, 
#          treeheight_row=0, fontsize_col = 1,
#          fontsize_row = 1,
#          cluster_rows=FALSE, 
#          cluster_cols=FALSE, 
#         annotation_row = annotCol,
#         annotation_col = annotCol,
#         annotation_colors=annotation_colors,
#          clustering_method="ward.D2",
#          color=colorRampPalette(c("blue", "white", "orange"))(50),main=cellLine)

# save_pheatmap_pdf(k, filename=paste0(cellLine,"_covMatrix.pdf"), width=9, height=7)

```
