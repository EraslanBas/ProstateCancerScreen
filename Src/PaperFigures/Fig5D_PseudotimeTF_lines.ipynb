# Fig 5D — TF motif accessibility across pseudotime (rising vs falling)
Based on **08_PlotTFActivity Part C** (line view). Coefficient trajectory across pseudotime segments (root NE=1 → Differentiated tip) of the top 10 **rising** (solid) and top 10 **falling** (dashed) TFs by Spearman trend. Coefficient = TF motif accessibility of each segment vs. the NE root; the full heatmap of all significant TFs is Fig S5E.

```python
from paperfig_style import *
import numpy as np, pandas as pd, matplotlib.pyplot as plt
```

```python
# --- 08_PlotTFActivity Part C line view (cell 12), replicated ----------------
pcoef = load_matrix('PseudotimeTFActivity_coef.csv')
tr = pd.read_csv(os.path.join(CSV_DIR, 'PseudotimeTFActivity_trend.csv'))
tr = tr[tr['nSigSeg'] >= 1]
seg = sorted(pcoef.columns, key=lambda c: int(c.replace('PT_seg', '')))
pcoef = pcoef[seg]; x = np.arange(1, len(seg)+1)
riseTF = [g for g in tr.sort_values('trend', ascending=False)['TF'] if g in pcoef.index][:10]
fallTF = [g for g in tr.sort_values('trend')['TF'] if g in pcoef.index][:10]
fig, ax = plt.subplots(figsize=(6.0, 3.4))
ax.axhline(0, color='grey', lw=0.6)
cmap = plt.get_cmap('tab20')
for i, g in enumerate(riseTF):
    ax.plot(x, pcoef.loc[g].values, '-',  lw=1.0, color=cmap(i % 20), label=f'{tf_label(g)} (rise)')
    ax.plot(x, pcoef.loc[g].values, '.',  ms=3, color=cmap(i % 20))
for i, g in enumerate(fallTF):
    ax.plot(x, pcoef.loc[g].values, '--', lw=1.0, color=cmap((i+10) % 20), label=f'{tf_label(g)} (fall)')
    ax.plot(x, pcoef.loc[g].values, '.',  ms=3, color=cmap((i+10) % 20))
ax.set_xticks(x)
ax.set_xlabel('pseudotime segment (root NE = 1 \u2192 Differentiated tip)')
ax.set_ylabel('Relative motif accessibility compared to the root')
ax.set_title('Top TFs changing across pseudotime', fontsize=8)
ax.legend(fontsize=4.5, ncol=2, loc='center left', bbox_to_anchor=(1.005, 0.5),
          handlelength=1.4, labelspacing=0.25)
savepanel(fig, 'Fig5D_PseudotimeTF_lines')
```
